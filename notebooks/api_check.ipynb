```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("EIA_API_KEY")

BASE_URL = "https://api.eia.gov/v2"
```

```python
response = requests.get(
    f"{BASE_URL}/electricity/rto/",
    params={
        "api_key": API_KEY
    },
    timeout=30
)

print(response.status_code)
print(response.json())
```

```python
response = requests.get(
    f"{BASE_URL}/electricity/rto/region-data/",
    params={
        "api_key": API_KEY
    },
    timeout=30
)

metadata = response.json()

metadata
```

```python
respondent_url = (
    f"{BASE_URL}/electricity/rto/region-data/"
    "facet/respondent/"
)

response = requests.get(
    respondent_url,
    params={"api_key": API_KEY},
    timeout=30
)

response.raise_for_status()

respondents = response.json()

for item in respondents["response"]["facets"][:20]:
    print(item)
```

```python
type_url = (
    f"{BASE_URL}/electricity/rto/region-data/"
    "facet/type/"
)

response = requests.get(
    type_url,
    params={"api_key": API_KEY},
    timeout=30
)

response.raise_for_status()

types = response.json()

for item in types["response"]["facets"]:
    print(item)
```
